```python
import sqlite3

import polars as pl
```

```python

with sqlite3.connect("data.sqlite") as conn:
    conn.execute("PRAGMA journal_mode=OFF")
    conn.execute("PRAGMA synchronous=OFF")
    conn.execute("PRAGMA foreign_keys=OFF")

    df = pl.read_database(
        "SELECT strategy,COUNT(*) cnt FROM aggregate_prizes WHERE ttl_reward >= ttl_payed GROUP BY strategy ",
        conn
    )
```

```python
df = df.drop("ratio")

tmp = df.with_columns(
    ((pl.col("cnt") / 750_000.0) * 100).alias("ratio")
)
print(tmp)

```
